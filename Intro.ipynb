# Quick Reference: NER and SpaCy

## What is NER?
**Named Entity Recognition (NER)** is a Natural Language Processing technique (not a library or model) that identifies and classifies named entities in text into predefined categories like Person, Organization, Location, etc.

## What is SpaCy?
**SpaCy** is an open-source Python library for Natural Language Processing - it provides tools and pre-trained models for various NLP tasks including tokenization, POS tagging, and NER.

## What is en_core_web_sm?
**en_core_web_sm** is a pre-trained statistical model (not a library) that comes with SpaCy - it contains trained weights and neural network parameters for processing English text and recognizing 18 default entity types.

## What We're Doing in This Project

### Using Pre-trained Model:
- We'll use SpaCy's **en_core_web_sm model** to understand how NER works
- This model recognizes general entities like PERSON, ORG, DATE, MONEY

### Training Custom Model:
- We'll train our **own custom NER model** using SpaCy library
- Our custom model will recognize resume-specific entities: NAME, DESIGNATION, SKILLS, DEGREE, COLLEGE_NAME, COMPANIES_WORKED_AT, GRADUATION_YEAR, EMAIL
- We use SpaCy's training pipeline to create this custom model

### In Summary:
- **SpaCy** = Library (provides tools and framework)
- **en_core_web_sm** = Pre-trained Model (general purpose, 18 entity types)
- **Our Custom Model** = New Model we'll build (resume-specific, 9 entity types)
- **NER** = Technique/Task (what we're trying to accomplish)

## SpaCy's Default Named Entity Types

```python
import spacy

# Load the pre-trained model
print("\n" + "="*70)
print("LOADING SPACY MODEL")
print("="*70)
nlp = spacy.load("en_core_web_sm")
print("Model loaded successfully: en_core_web_sm")

# Display all default entity types
print("\n" + "="*70)
print("DEFAULT ENTITY TYPES IN SPACY en_core_web_sm MODEL")
print("="*70)
```

```python
#  Entity types with descriptions
entity_types = {
    "PERSON": "People, including fictional characters",
    "NORP": "Nationalities, religious or political groups",
    "FAC": "Buildings, airports, highways, bridges, etc.",
    "ORG": "Companies, agencies, institutions, etc.",
    "GPE": "Countries, cities, states (Geo-Political Entities)",
    "LOC": "Non-GPE locations, mountain ranges, bodies of water",
    "PRODUCT": "Objects, vehicles, foods, etc. (not services)",
    "EVENT": "Named hurricanes, battles, wars, sports events, etc.",
    "WORK_OF_ART": "Titles of books, songs, movies, paintings, etc.",
    "LAW": "Named documents made into laws",
    "LANGUAGE": "Any named language",
    "DATE": "Absolute or relative dates or periods",
    "TIME": "Times smaller than a day",
    "PERCENT": "Percentage, including '%'",
    "MONEY": "Monetary values, including unit",
    "QUANTITY": "Measurements, as of weight or distance",
    "ORDINAL": "First, second, third, etc.",
    "CARDINAL": "Numerals that do not fall under another type"
}

print("\nTotal Entity Types: 18\n")
for entity, description in entity_types.items():
    print(f"{entity:15} : {description}")
```

# SpaCy NER Model Architecture

"""
### Internal Architecture of SpaCy NER:

1. TEXT INPUT
   "Apple Inc. was founded by Steve Jobs"

2. TOKENIZATION
   ["Apple", "Inc.", "was", "founded", "by", "Steve", "Jobs"]

3. NEURAL NETWORK LAYERS
   - Input Layer: Token embeddings
   - Hidden Layers: Multiple CNN or Transformer layers
   - Output Layer: Entity type probabilities for each token

4. ENTITY RECOGNITION
   - Uses IOB (Inside-Outside-Begin) tagging scheme
   - B-ORG: Beginning of Organization entity
   - I-ORG: Inside/Continuation of Organization entity
   - O: Outside any entity

5. OUTPUT
   Apple Inc. -> ORG
   Steve Jobs -> PERSON

### Training Process:
- Pre-trained on OntoNotes 5.0 corpus
- Contains ~2 million words of annotated text
- Trained on news articles, web text, broadcast conversations
- Learns patterns and context for entity recognition
"""

```python
 # Example 1: Simple Text Processing

print("\n" + "="*70)
print("EXAMPLE 1: EXTRACTING ENTITIES FROM SIMPLE TEXT")
print("="*70)


text1 = "Apple Inc. was founded by Steve Jobs in Cupertino, California in 1976."
print(f"\nInput Text:\n{text1}")

# Process text with SpaCy
doc1 = nlp(text1)

# Extract and display entities
print("\nEntities Found:")
print(f"{'Entity Text':<25} {'Label':<15} {'Description'}")
print("-"*70)
for ent in doc1.ents:
    print(f"{ent.text:<25} {ent.label_:<15} {spacy.explain(ent.label_)}")
```

```python
# Example 2: News Article Text

print("\n" + "="*70)
print("EXAMPLE 2: EXTRACTING ENTITIES FROM NEWS ARTICLE")
print("="*70)

text2 = """
Elon Musk announced that Tesla will open a new factory in Austin, Texas
next month. The facility will create 5,000 jobs and cost $1.1 billion to build.
The project was approved by the Texas government last week.
"""

print(f"Input Text:\n{text2}")

doc2 = nlp(text2)

print("\nEntities Found:")
print(f"{'Entity Text':<25} {'Label':<15} {'Start':<8} {'End':<8}")
print("-"*70)
for ent in doc2.ents:
    print(f"{ent.text:<25} {ent.label_:<15} {ent.start_char:<8} {ent.end_char:<8}")
```

```python
 ## Example 3: Resume Text

print("\n" + "="*70)
print("EXAMPLE 3: APPLYING NER TO RESUME TEXT")
print("="*70)

resume_text = """
Rahul Sharma
Senior Software Engineer
Bengaluru, Karnataka

Email: rahul.sharma@email.com
Phone: +91-9876543210

PROFESSIONAL SUMMARY:
Experienced software engineer with 5 years at Google and Microsoft.
Expertise in Python, Java, Machine Learning, and AWS.

EDUCATION:
B.Tech in Computer Science
Indian Institute of Technology, Mumbai
Graduated: 2018
"""

print(f"Sample Resume Text:\n{resume_text}")

doc3 = nlp(resume_text)

print("\nEntities Extracted by Pre-trained Model:")
print(f"{'Entity Text':<35} {'Label':<15}")
print("-"*70)
for ent in doc3.ents:
    print(f"{ent.text:<35} {ent.label_:<15}")
```
